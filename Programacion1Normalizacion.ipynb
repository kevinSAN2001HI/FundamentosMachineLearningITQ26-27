**Normalización:** Proceso en el cual convertimos nuestros datos en un rango de [0, 1]. A este proceso se lo denomina homotencia.

**Formula:**
$x^n= x_i - min(x_i) / max(x_i) - min(x_i)$

```python
#Carga de librerias 
from sklearn import preprocessing
import numpy as np 
```

**Sklearn** Biblioteca de aprendizaje automático (machine learning) de código abierto para el lenguaje de programación Python.

**Preprocessing** Modulo de preprocesamiento util para transformar datos sucios o desordenados en información limpia y estructurada. Entrega datos limpios para ser utilizada por un algoritmo o modelo aoartir de procesos como la normalización y estandarización.

**Numpy**  Biblioteca oficial, diseñada para la computación científica y el manejo eficiente de matrices y arreglos multidimensionales.

**np** al utilizar *as* le asignamos un alias a la biblioteca nummpy y cada vez que utilicemos la funcion solo la llamamos con su alias asignada.

```python
# Carga de datos 
X_train = np.array([[3., 1., 2.],
                    [0., 2., 5.],
                    [2., 3., -2.]])
                   
print(X_train)
```

**Array:**   Es una estructura de datos que sirve para almacenar múltiples valores o elementos dentro de una sola variable. En este caso almacenada en unamatriz de 3x3.


```python
# Limpieza de datos
# Normalización
normalizer = preprocessing.MinMaxScaler() 
X_train_norm = normalizer.fit_transform(X_train)
```

***Preprocessing*** Es una caja de herramientas de la librería utilizada "*scikit-learn*". La cual contiene una colección de funciones y algoritmos diseñados para limpiar, transformar y preparar los datos brutos antes de entregárselos a un modelo de Machine Learning.

**MinMaxScaler:** Objeto se encargará de calcular y aplicar la fórmula de normalización $x^n= x_i - min(x_i) / max(x_i) - min(x_i)$ . Transformará tus datos para que el valor más pequeño sea 0 y el más grande sea 1.


***Normalizer*** Es una variable el encargado de calcular los valores mínimos y máximos de tus datos y transformarlos al rango 0-1.


**Fit:** Analiza la variable para encontrar y memorizar el valor mínimo y el valor máximo de cada columna.

**Transform:** Con los mínimos y máximos calculados por "*Fit*", aplica la fórmula a todos los datos de de la variable, transformándolos al rango 0-1. 


```python
# Salida de resultados
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos␣ "fit" y "transform" por separado 

X_train_2 = np.copy(X_train) 

# Crea un objeto "fitted_normalizer" llamando al método "fit" 

f = normalizer.fit(X_train) 
X_train_norm_2 = f.transform(X_train) 

# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2" 
# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales 

print(np.array_equiv(X_train_norm, X_train_norm_2))
```

**Copy:** Es una función de NumPy que realiza una copia profunda del arreglo. Permite crear una copia identica sin alterar la varible. Utilizada para experimentar y alterar datos y ver su funcionamiento.

**Array_equiv:** Es la función que compara dos arreglos para ver si son equivalentes. Devuelve un único valor: *True* si son iguales, o *False* si hay alguna diferencia.
