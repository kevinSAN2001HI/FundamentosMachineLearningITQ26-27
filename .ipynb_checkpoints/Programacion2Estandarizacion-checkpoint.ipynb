**Normalización:** Proceso en el cual transformamos la distribucion de valores del atributo para que estos adquieran una media y una desviación estandar típica determinada.

**Formula:**
$x^s= x_i - media(x_i) / desv(x_i)$

```python
from sklearn import preprocessing 
import numpy as np
```

```python
# Carga de datos
X_train = np.array([[ 3., 1., -2.], 
                    [ 2., 0., 1.], 
                    [ 1., 1., -1.]])
```

```python
# Limpieza de datos: 
#Estandarización
standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento 
X_train_std=standardizer.fit_transform(X_train) 
print(X_train_std)
```

**Standar** Módulo que pertenece a la librería scikit-learn utilizado para el preprocesamiento de datos como la limpieza, transformación y preparación de los datos antes de introducirlos en un modelo.

**Scaler** Es una clase dentro del módulo preprocessing el cual implementa la estandarización.

```python
X_train_2 = np.array([[ 3., 1., -2.], 
                    [ 2., 0., 1.], 
                    [ 1., 1., -1.]])
```

```python
# Aplicamos los métodos "fit" y "transform" por separado 
f=preprocessing.StandardScaler() 
f=f.fit(X_train)
X_train_2_std = f.transform(X_train_2) 
print(X_train_2_std) 
```

```python
# Comprobar si las matrices estandarizadas son iguales

print(np.array_equiv(X_train_std, X_train_2_std))
```
